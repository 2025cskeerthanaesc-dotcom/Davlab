```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
from sklearn.metrics import precision_score,recall_score,f1_score
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv(r"C:\Users\Keerthana Reddy\Downloads\archive\StudentPerformanceFactors.csv")
print(df.head())
print(df.columns)
```

```python
df=df.dropna(subset=["Hours_Studied","Exam_Score"])
df["Pass"]=(df["Exam_Score"]>=60).astype(int)
print(df[["Hours_Studied","Exam_Score","Pass"]].head(10))
print(df["Pass"].value_counts())
```

```python
x=df[["Hours_Studied"]]
y=df["Pass"]
print("Input(x):")
print(x.head())
print("Output(y):")
print(y.head())
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42,stratify=y)
print("Training  data;",x_train.shape)
print("Testing  data;",x_test.shape)
```

```python
model=LogisticRegression()
model.fit(x_train,y_train)
print("Model trained successfully")
```

```python
y_pred=model.predict(x_test)
print("actual Values:")
print(y_test.values)
print("Predicted Values")
print(y_pred)
```

```python
print("Accuracy:",accuracy_score(y_test,y_pred))
```

```python
cm=confusion_matrix(y_test,y_pred)
print("Confusion Matrix")
print(cm)
```

```python
print("Precision:",precision_score(y_test,y_pred,zero_division=0))
print("Recall:",recall_score(y_test,y_pred,zero_division=0))
print("F1-score:",f1_score(y_test,y_pred,zero_division=0))

```
